```python
from IPython.core.debugger import coloransi
class Pessoa:
  #Construtor
  def __init__(self, nome, idade, Cargo):
    self.nome = nome
    self.idade = idade
    self.cargo = Cargo



# Metodos
  def informacoes(self):
    print(f'Nome:  {self.nome}')
    print(f'Idade: {self.idade}' )
    print(f'Cargo: {self.cargo}' )


  def ligar(self):
    if not self.ligado:
      self.ligado = True
      print('O carro foi Ligado')
    else:
      print('o Carro ja estava ligado')

  def promover(self, novo_cargo):
    print(f'{self.nome} foi promovido(a) para a nova  funcao de {novo_cargo} ')



  def atualizar_idade(self, nova_idade):
    if nova_idade > self.idade:
      print(f'atualizando a idade de {self.idade} para {nova_idade}')
    else:
      print(f'A nova idade tem que ser maior que a antiga ')


# Cadastrar a primeira casa(=Instancias)

colaborador1 = Pessoa('Lopes', 27, 'Assistente Junior')
colaborador2 = Pessoa('Luzia', 27, 'DEV Junior')


colaborador1.informacoes()
colaborador1.promover('AssisteDevnte Plene')
colaborador2.atualizar_idade(55)
```
