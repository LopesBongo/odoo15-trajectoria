```python
class Pessoa:
  def __init__(self,nome,idade):
    self.nome = nome
    self.idade = idade


  def apresentacao(self):
    print(f'Olá o seu Nome é {self.nome} e tenho {self.idade} de idade')

class Funcionario(Pessoa):

  def __init__(self,nome,idade,cargo):
    super().__init__(nome,idade)#função embutida do pai

    self.cargo = cargo

  def trabalhar(self):
    print(f'{self.nome} Está trabalhando como {self.cargo}')


  def apresentacao(self):
    print(f'Olá o seu Nome é {self.nome} e tenho {self.idade} de idade e meu cargo é {self.cargo}')

class Cliente(Pessoa):
  def __init__(self,nome,idade, saldo):
    super().__init__(nome,idade)
    self.saldo = saldo

  def comprar(self, valor_compra):

    if valor_compra <= self.saldo:

      self.saldo -= valor_compra
      print(f'olá {self.nome}! A sua compra de {valor_compra} foi Aprovada! seu saldo atual é de: {self.saldo}')
    else:
      print(f'olá {self.nome}! o seu Saldo foi insuficiente para compra')


f1 = Funcionario('Lopes',38, 'Gerente de conta')
#f1.apresentacao()

c1 = Cliente('Luzia',16, 300)
c2 = Cliente('Lopes',38, 50)
#c1.apresentacao()
c1.comprar(200)
c2.comprar(30)


```
