```python
#Sistema de escola

class Escola():
    def __init__(self, nome,idade, status):
        self.nome = nome
        self.idade = idade
        self.status = status

    def apresentar(self):
       print(f'Olá, meu nome é {self.nome}, tenho {self.idade} anos , o meu estado é {self.status}')

    def verficar_status(self):
      print(f'Status: {"ATIVO" if self.status else "INATIVO"}')


class Aluno(Escola):
    def __init__(self, nome,idade, ano,status):

      super().__init__(nome,idade,status)
      self.ano = ano

    def apresentar(self):
      #super().apresentar()
      print(f'Olá, meu nome é {self.nome}, tenho {self.idade} anos e estudo no {self.ano} ano')


class Professor(Escola):
    def __init__(self, nome, idade, materia,status):
        super().__init__(nome, idade, status)
        self.materia = materia
    def apresentar(self):
        print(f'Olá, meu nome é {self.nome}, tenho {self.idade} anos e leciono a materia de {self.materia}')

class assistente(Escola):
  def __init__(self, nome, idade, bloco,status):

    super().__init__(nome, idade, status)
    self.bloco = bloco

  def apresentar(self):
    print(f'Olá, meu nome é {self.nome}, tenho {self.idade} anos e trabalho no: {self.bloco}')

a1 = Aluno(nome='Pedro', idade=18, ano=1,status=True)

p1 = Professor(nome='João', idade= 40, materia='Matematica',status=True)
as1 = assistente(nome='Maria',  idade=30, bloco='Bloco A',status=False)

a1.apresentar()
a1.verficar_status()


p1.apresentar()
p1.verficar_status()


as1.apresentar()
as1.verficar_status()
```
