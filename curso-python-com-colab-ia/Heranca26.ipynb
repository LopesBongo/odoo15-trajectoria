```python
#Herança Multipla

#classes pai

class Predador():
  def cacando(self):
    print('Este animal esta Caçando!')

class Presa():
  def fugindo(self):
    print('Este animal esta sendo  caçado!')


#Classes Filhos

class Coelho(Presa):
  pass

class Tigre(Predador):
  pass

class Golfinho(Predador, Presa):
  pass

Coelho1 = Coelho()
Tigre1 = Tigre()
Golfinho1 = Golfinho()



Coelho1.fugindo()
Tigre1.cacando()
Golfinho1.cacando()
Golfinho1.fugindo()
```
