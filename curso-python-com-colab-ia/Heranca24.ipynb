```python
class Animal:
  def __init__(self,nome,cor,especie):
    self.nome = nome
    self.cor = cor
    self.especie = especie

  def apresentacao(self):
    print(f'Eu Sou o {self.especie} chamado {self.nome}')

class Gato(Animal):

  def emitir_som(self):
    print('Miau!')

  def arranhar(self):
    print('o gato esta arranhando')

class Cachorro(Animal):
  def emitir_som(self):
    print('AU AU AU.....!')



class Elefante(Animal):
   def emitir_som(self):
    print('Pruuuuuuuuuuu.....!')





gato1 = Gato('Felix','Branco', 'Siamese')
gato1.apresentacao()
gato1.emitir_som()
gato1.arranhar()


Cachorro1 = Cachorro('Russo','Preto', 'Pastor Alemão')
Cachorro1.apresentacao()
Cachorro1.emitir_som()


Elefante1 = Elefante('Fred','Marro','Asiatico')
Elefante1.apresentacao()
Elefante1.emitir_som()
```
